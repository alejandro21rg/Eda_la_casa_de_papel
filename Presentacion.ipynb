Impacto internacional de *La Casa de Papel* y la influencia de su triunfo en las series españolas.

## Tema

Este EDA analiza el impacto internacional de la serie española *La Casa de Papel* utilizando datos procedentes de Netflix e IMDb.

A través de métricas como visualizaciones globales, permanencia en rankings, puntuaciones y número de votos, se estudia el alcance internacional de la serie y su posicionamiento frente a otras producciones españolas e internacionales.

Además, se realiza una comparación entre series españolas anteriores y posteriores a 2017, año de estreno de *La Casa de Papel*, con el objetivo de analizar si su éxito internacional supuso un punto de inflexión para la percepción y popularidad global de las series españolas.

## Data

Para el análisis se utilizan distintos datasets procedentes de Netflix e IMDb.

#### Netflix
- `most-popular_country_weekly.tsv`
- `most-popular_global_alltime.tsv`
- `most-popular_global_weekly.tsv`

#### IMDb
- `title.basics.tsv`
- `title.ratings.tsv`

#### Variables principales
- `weekly_views`
- `cumulative_weeks_in_top_10`
- `averageRating`
- `numVotes`
- `genres`
- `startYear`

# Hipótesis

## Hipótesis

### Hipótesis principal

*La Casa de Papel* se consolidó como la serie española con mayor impacto internacional gracias a su entrada en Netflix, alcanzando niveles de visualización, permanencia en rankings y reconocimiento global sin precedentes para una producción española.

    - Top 10, series más vistas en Netflix de no habla inglesa.
    - Top 8, global.
    - Serie española más vista en diferentes paises: Francia, Argentina, EEUU... 
    - Serie españolas que más tiempo ha estado dentro del top 10 a nivel global
    - Serie espaola mejor valorada

### Hipótesis secundarias

Además, su éxito internacional supuso un punto de inflexión para la percepción global de las series españolas, aumentando la visibilidad y aceptación internacional de nuevas producciones españolas a partir de 2017.

    - Aumento de las visualizaciones de series españolas a partir de ese año.
    - Incremento de la valoración de estas misma.



# Metodologia

Se realizó un proceso de limpieza y transformación de datos utilizando Python y Pandas.

Posteriormente:
- se filtraron únicamente series españolas,
- se unificaron datasets de Netflix e IMDb,
- se compararon métricas antes y después de 2017,
- y se analizaron rankings globales, visualizaciones y valoraciones.
