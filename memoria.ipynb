# 1. Introducción

En los últimos años, las plataformas de streaming han transformado la industria audiovisual, modificando la forma en la que las producciones cinematográficas y televisivas son distribuidas y consumidas a nivel mundial. Entre estas plataformas, Netflix ha desempeñado un papel fundamental en la internacionalización de contenidos procedentes de países no angloparlantes, permitiendo que producciones locales alcancen una visibilidad global sin precedentes.

Dentro de este contexto, *La Casa de Papel* se convirtió en uno de los mayores fenómenos audiovisuales internacionales surgidos en España. Tras su incorporación al catálogo de Netflix en 2017, la serie experimentó un crecimiento exponencial en popularidad, situándose entre las producciones más vistas y reconocidas de la plataforma. Su impacto trascendió el ámbito nacional y europeo, logrando posicionarse en rankings internacionales y consolidándose como una de las series de habla no inglesa con mayor repercusión global.

Además de su éxito en términos de visualizaciones, *La Casa de Papel* alcanzó un elevado reconocimiento por parte de la audiencia internacional, reflejado tanto en su permanencia dentro de los rankings globales de Netflix como en sus valoraciones en IMDb. Este fenómeno no solo impulsó la popularidad de la propia serie, sino que también contribuyó al aumento de la visibilidad internacional de las producciones españolas.

El presente proyecto desarrolla un Análisis Exploratorio de Datos (EDA) con el objetivo de estudiar el impacto internacional de *La Casa de Papel* mediante datos procedentes de Netflix e IMDb. Para ello, se analizarán métricas relacionadas con visualizaciones, rankings globales, permanencia en el Top 10, valoraciones y número de votos, comparando el comportamiento de la serie frente a otras producciones españolas e internacionales.

Asimismo, se investigará si el éxito internacional de *La Casa de Papel* supuso un punto de inflexión para la percepción global de las series españolas, favoreciendo el incremento de nuevas producciones nacionales con presencia internacional a partir de 2017.

Finalmente, este análisis permitirá comprender cómo una producción española logró competir dentro de un mercado dominado tradicionalmente por contenidos angloparlantes, convirtiéndose en un fenómeno global impulsado por las plataformas de streaming.

# 2. Objetivos

## 2.1 Objetivo principal

El objetivo principal de este proyecto es analizar el impacto internacional de *La Casa de Papel* a través de datos procedentes de Netflix e IMDb, con el fin de determinar si la serie se consolidó como la producción española con mayor repercusión nacional y global dentro de las plataformas de streaming.

Para ello, se estudiarán métricas relacionadas con visualizaciones, permanencia en rankings, popularidad y valoraciones internacionales, comparando el comportamiento de la serie con otras producciones españolas y con algunas de las series internacionales más relevantes de Netflix.

---

Para alcanzar el objetivo principal se plantean los siguientes objetivos específicos:

- Identificar si *La Casa de Papel* presenta métricas significativamente superiores.

### 🌍 Analizar el alcance internacional de la serie

- Estudiar la presencia de *La Casa de Papel* en rankings globales de Netflix.
- Evaluar su permanencia dentro del Top 10 internacional.

### ⭐ Evaluar el reconocimiento obtenido en IMDb

- Analizar la valoración media de la serie.
- Estudiar el volumen de votos recibidos.

### 📈 Analizar el impacto de la serie en las producciones españolas posteriores

- Estudiar la evolución de las series españolas a partir de 2017.

- Evaluar posibles cambios en popularidad y valoración tras el éxito de la serie.

# 4. Descripción de los datos

Para el desarrollo de este proyecto se utilizaron distintos datasets procedentes de Netflix e IMDb. Estos conjuntos de datos permiten analizar tanto el rendimiento de las series dentro de la plataforma de streaming como su reconocimiento y valoración por parte de la audiencia internacional.

La combinación de ambas fuentes proporciona una visión más completa del impacto internacional de *La Casa de Papel*, permitiendo estudiar simultáneamente métricas de popularidad, visualizaciones y valoración crítica.

---

# 4.1 Fuentes de datos

## 📺 Netflix

Los datasets de Netflix contienen información relacionada con rankings globales, permanencia en el Top 10 y número de visualizaciones de las producciones disponibles en la plataforma.

### Datasets utilizados

| Dataset | Descripción |
|---|---|
| `most-popular_country_weekly.tsv` | Rankings semanales de series por país |
| `most-popular_global_alltime.tsv` | Series más vistas globalmente en Netflix |
| `most-popular_global_weekly.tsv` | Rankings globales semanales de Netflix |

---

## ⭐ IMDb

IMDb proporciona información relacionada con la valoración y popularidad de las producciones audiovisuales a nivel internacional.

### Datasets utilizados

| Dataset | Descripción |
|---|---|
| `title.basics.tsv` | Información general de títulos audiovisuales |
| `title.ratings.tsv` | Valoraciones medias y número de votos |

---

# 4.2 Variables principales analizadas

A partir de los datasets seleccionados se trabajó con distintas variables relevantes para el análisis.

| Variable | Descripción |
|---|---|
| `weekly_views` | Número de visualizaciones semanales |
| `cumulative_weeks_in_top_10` | Semanas acumuladas dentro del Top 10 |
| `averageRating` | Valoración media en IMDb |
| `numVotes` | Número total de votos recibidos |
| `genres` | Géneros asociados a la serie |
| `startYear` | Año de estreno |
| `country_name` | País del ranking semanal |
| `weekly_rank` | Posición semanal en rankings |

---

# 4.3 Selección de los datos

El análisis se centró principalmente en producciones españolas disponibles en Netflix, utilizando *La Casa de Papel* como caso principal de estudio.

Para contextualizar los resultados también se realizaron comparaciones con:

- Otras series españolas.
- Producciones internacionales.
- Series de habla no inglesa presentes en rankings globales.

Además, se consideró especialmente el periodo posterior a 2017, año en el que *La Casa de Papel* fue incorporada al catálogo internacional de Netflix.

---

# 4.4 Limitaciones de los datos

A pesar de la utilidad de los datasets utilizados, existen algunas limitaciones que deben tenerse en cuenta durante el análisis.

## Principales limitaciones

- Netflix no proporciona acceso completo a todas las métricas internas de visualización.
- Algunos rankings corresponden únicamente a determinados periodos temporales.
- IMDb refleja valoraciones de usuarios, por lo que existe un componente subjetivo.
- No todas las series españolas disponen del mismo volumen de datos o exposición internacional.

No obstante, los datasets seleccionados permiten obtener una aproximación sólida al impacto internacional de *La Casa de Papel* y al comportamiento global de las producciones españolas dentro de las plataformas de streaming.
