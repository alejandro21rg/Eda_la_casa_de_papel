# 1. Introducción

En los últimos años, las plataformas de streaming han transformado la industria audiovisual, modificando la forma en la que las producciones cinematográficas y televisivas son distribuidas y consumidas a nivel mundial. Entre estas plataformas, Netflix ha desempeñado un papel fundamental en la internacionalización de contenidos procedentes de países no angloparlantes, permitiendo que producciones locales alcancen una visibilidad global sin precedentes.

Dentro de este contexto, *La Casa de Papel* se convirtió en uno de los mayores fenómenos audiovisuales internacionales surgidos en España. Tras su incorporación al catálogo de Netflix en 2017, la serie experimentó un crecimiento exponencial en popularidad, situándose entre las producciones más vistas y reconocidas de la plataforma. Su impacto trascendió el ámbito nacional y europeo, logrando posicionarse en rankings internacionales y consolidándose como una de las series de habla no inglesa con mayor repercusión global.

Además de su éxito en términos de visualizaciones, *La Casa de Papel* alcanzó un elevado reconocimiento por parte de la audiencia internacional, reflejado tanto en su permanencia dentro de los rankings globales de Netflix como en sus valoraciones en IMDb. Este fenómeno no solo impulsó la popularidad de la propia serie, sino que también contribuyó al aumento de la visibilidad internacional de las producciones españolas.

El presente proyecto desarrolla un Análisis Exploratorio de Datos (EDA) con el objetivo de estudiar el impacto internacional de *La Casa de Papel* mediante datos procedentes de Netflix e IMDb. Para ello, se analizarán métricas relacionadas con visualizaciones, rankings globales, permanencia en el Top 10, valoraciones y número de votos, comparando el comportamiento de la serie frente a otras producciones españolas e internacionales.

Asimismo, se investigará si el éxito internacional de *La Casa de Papel* supuso un punto de inflexión para la percepción global de las series españolas, favoreciendo el incremento de nuevas producciones nacionales con presencia internacional a partir de 2017.

Finalmente, este análisis permitirá comprender cómo una producción española logró competir dentro de un mercado dominado tradicionalmente por contenidos angloparlantes, convirtiéndose en un fenómeno global impulsado por las plataformas de streaming.

Argentina

```python
series_argentina = series_espanolas[ series_espanolas["country_name"] == "Argentina"]

series_argentina.head(10)
```

```python
series_argentina_ord = series_argentina.sort_values(by="cumulative_weeks_in_top_10", ascending=False)

series_argentina_ord.head(10)
```

```python
# Top 10
top10 = (
    series_argentina
    .groupby("show_title")["cumulative_weeks_in_top_10"]
    .max()
    .sort_values(ascending=False)
    .head(10)
)

plt.style.use("default")

# Figura
plt.figure(figsize=(12,7))

# Barras horizontales
bars = plt.barh(
    top10.index,
    top10.values,
    color="#E50914" 
)

# Invertimos para que el mayor quede arriba
plt.gca().invert_yaxis()

# Títulos
plt.title(
    "Top Series españolas en Argentina",
    fontsize=20,
    fontweight="bold"
)

# Añadir números al final de cada barra
for bar in bars:
    width = bar.get_width()
    
    plt.text(
        width + 0.3,
        bar.get_y() + bar.get_height()/2,
        f"{int(width)}",
        va="center",
        fontsize=11
    )

plt.xlabel(
    "Semanas acumuladas en Top 10",
    fontsize=12
)

# Ajustar diseño
plt.tight_layout()

plt.show()
```

Para ello vamos a importar dos csv de IMDb donde lo analizaremos


```python
# Los dos csv para despues agruparlos con un merge
df_idmb_title_basics = pd.read_csv("Código/data/title.basics.tsv", sep="\t")
df_idmb_title_ratings = pd.read_csv("Código/data/title.ratings.tsv", sep="\t")
```

Vemos los dos csv

```python
# DF con los datos de las series mas el tconts que tiene el otro df
df_idmb_title_basics
```

```python
# Puntuacion, votos y tconst
df_idmb_title_ratings
```

```python
# El merge para juntar los csv
imdb_completo = pd.merge(df_idmb_title_basics,df_idmb_title_ratings, on="tconst")
```

```python
# Del DF final me quedo con las series
series_imdb = imdb_completo[imdb_completo["titleType"] == "tvSeries"]
```

```python
series_imdb
```

```python
# Para que se vean titulo, genero, nota y numero de votos

series_final = series_imdb[[
    "primaryTitle",
    "startYear",
    "genres",
    "averageRating",
    "numVotes"
]]
```

```python
series_final
```

Creo un rankin global (me servirá más tarde)

```python

ranking_global = (
    series_final.sort_values(
        by="numVotes",
        ascending=False
    )
    .reset_index(drop=True)
)
ranking_global
```

Vamos a compararlas con otros paises Europeos

```python


# Función para contar producciones por país
def country_growth(code):
    
    country = crecimiento_esp[
        crecimiento_esp["production_countries"] == f"['{code}']"
    ]

    growth = country.groupby("release_year").size()

    growth = growth[
        (growth.index >= 2014) &
        (growth.index <= 2021)
    ]

    return growth

# Datos países
spain = country_growth("ES")
france = country_growth("FR")
germany = country_growth("DE")
italy = country_growth("IT")
uk = country_growth("GB")

# Figura
plt.figure(figsize=(15,8))

# España destacada
plt.plot(
    spain.index,
    spain.values,
    linewidth=4,
    marker="o",
    label="España"
)

# Otros países
plt.plot(
    france.index,
    france.values,
    linewidth=2.5,
    marker="o",
    label="Francia"
)

plt.plot(
    germany.index,
    germany.values,
    linewidth=2.5,
    marker="o",
    label="Alemania"
)

plt.plot(
    italy.index,
    italy.values,
    linewidth=2.5,
    marker="o",
    label="Italia"
)

plt.plot(
    uk.index,
    uk.values,
    linewidth=2.5,
    marker="o",
    color="#1D3557",   # Color Reino Unido
    label="Reino Unido"
)

# Línea vertical La Casa de Papel
plt.axvline(
    x=2017,
    color="red",
    linestyle="--",
    linewidth=2
)


# Títulos
plt.title(
    "Evolución de producciones europeas en Netflix",
    fontsize=18,
    fontweight="bold"
)

plt.xlabel("Año", fontsize=12)
plt.ylabel("Número de producciones", fontsize=12)

plt.legend()

plt.grid(alpha=0.3)

plt.show()
```
